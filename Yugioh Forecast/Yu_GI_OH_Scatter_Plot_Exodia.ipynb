```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import csv 
```

```python
#read from dataset
df = pd.read_csv("yugioh_enriched.csv", usecols=['name','card_sets_set_name','card_sets_set_price'])
df.columns
```

```python
#Read top 5 rows
df.head(5)
```

```python
#Search row for specific card
df= df[df['name']=='Exodia the Forbidden One'].reset_index()

#Check top rows
df.head(15)
```

```python
#Generate Scatter Plot for Card Prices

fig, ax = plt.subplots(1,1, figsize=(40,20))

x= df['card_sets_set_name']
y= df['card_sets_set_price']

plt.scatter(x,y)

plt.title("Exodia Price: Across Sets")
plt.legend(["Price"])
plt.xlabel("Card Set")
plt.ylabel("Card Price")

plt.show()
```
