```python
#Lasso estimates sparse coeficients. It is useful in some circumstances due to its tendency to prefer solutions with fewer non-zero coefficients, 
#effectively reducing the number of features upon which the given solution is dependant
#The lasso estimate solves the minimization of the least-squares penalty with a||w||1 added, where a is a constant and ||w||1 is the L1 norm of the coefficient vector
```

```python
from sklearn import linear_model
reg = linear_model.Lasso(alpha=0.1)
reg.fit([[0, 0], [1, 1]], [0, 1])
reg.predict([[1, 1]])
```
