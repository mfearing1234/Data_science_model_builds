```python
strs = ['flower','flow','flight']

class Solution:
    def longestCommonPrefix(self, strs):
        n = len(strs)
        if n == 0:
            return ""
        min_length = float('inf')
        for s in strs:
            if len(s) < min_length:
                min_length = len(s)

        i = 0
        while i < min_length:
            for s in strs:
                if i == len(s) or s[i] != strs[0][i]:
                    return strs[0][:i]
            i += 1
        return strs[0][:i]
    
oop_solution = Solution()
oop_solution.longestCommonPrefix(strs)
```
