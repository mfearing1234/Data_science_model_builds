```python
# OO_LightSwitch

class LightSwitch():
    def __init__(self):
        self.switchIsOn = False

    def turnOn(self):
        # turn the switch on 
        self.switchIsOn = True

    def turnOff(self):
        # turn the switch off
        self.switchIsOn = False
ls = LightSwitch()
ls.turnOff()
```
