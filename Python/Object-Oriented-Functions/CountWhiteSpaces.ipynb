```python
x = 'I am cold'
class SpaceCounter:
    def count_spaces(self, sentence):
        return sentence.count(' ')


s = SpaceCounter()
print(s.count_spaces(x))
```

```python
sentence = 'I      go  to mall'

class space_counter:
    """Blueprint for counting spaces"""
    def count_spaces(self, x):
        spaces = 0
        for i in x:
            if i == ' ' and x[-1] != ' ':
                spaces = spaces + 1
        return spaces

s=space_counter
s.count_spaces(s,sentence)
```
