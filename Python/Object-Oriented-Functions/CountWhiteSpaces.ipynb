```python
sentence = 'I      go  to mall'

class space_counter:
    """Blueprint for counting spaces"""
    def count_spaces(self, x):
        spaces = 0
        for i in x:
            if i == ' ' and x[-1] != ' ':
                spaces = spaces + 1
        return spaces

s=space_counter
s.count_spaces(s,sentence)
```

```python
sentences = 'I go    swimmi ng'
class space_counter():
    def __init__(self):
        self.spaces = 0
        self.previous = False
    def total_spaces (self, sentence):
        for i in sentence:
            if i == ' ':
                self.spaces += 1
        return print(f"There are {self.spaces} spaces between the letters in the sentence.")
    def unique_space (self,sentence):
        for i in sentence:
            if i == ' ' and self.previous == False:
                self.spaces += 1
                self.previous = True
            elif i != ' ':
                self.previous = False

        return print(f"There are {self.spaces} spaces between the letters in the sentence.")
sc= space_counter()
sc.unique_space(sentences)
sc.total_spaces(sentences)
```
