```python
x = 'I am cold'

def count_spaces(sentence):
    spaces = 0
    for i in sentence:
        if i == ' ':
            spaces = spaces + 1
    print(spaces)


count_spaces(x)
```

```python
x = 'I am cold'
class SpaceCounter:
    def count_spaces(self, sentence):
        return sentence.count(' ')


s = SpaceCounter()
print(s.count_spaces(x))
```
