## Problem
Scenario: A factory makes two products, A and B.

Each unit of A yields $40 profit; each unit of B yields $30 profit.
Product A requires 2 hours of labor and 3 lbs of raw material per unit.
Product B requires 1 hour of labor and 2 lbs of raw material per unit.
Available: 100 labor hours, 180 lbs of raw material.
Demand cap: no more than 40 units of A can be sold.

```python
import pandas as pd
import numpy as np
import pulp as pl
```

# Generate Data Sets 

```python
dic_product_cost = {"Product_A": 40, "Product_B": 30}
dic_product_labor = {"Product_A": 2, "Product_B": 1}
dic_product_material = {"Product_A": 3, "Product_B": 2}
dic_available_labor = {"Labor_capacity": 100}
dic_available_material = {"Material_capacity": 180}
```

# Problem Definition

```python
#define the problem
problem = pl.LpProblem("Production_Planning", pl.LpMaximize)
```

## Decision Variables

```python
# Define Decision Variables
product_A = pl.LpVariable("Product_A", lowBound=0, cat='Integer')
product_B = pl.LpVariable("Product_B", lowBound=0, cat='Integer')
```

# Objective Function

```python
# Objective Function to maximize profit
problem += dic_product_cost["Product_A"] * product_A + dic_product_cost["Product_B"] * product_B, "Total_Profit"
```

# Constraints

```python
# Constraints
#Labor constraint
problem += dic_product_labor["Product_A"] * product_A + dic_product_labor["Product_B"] * product_B <= dic_available_labor["Labor_capacity"], "Labor_Constraint"
#Material constraint
problem += dic_product_material["Product_A"] * product_A + dic_product_material["Product_B"] * product_B <= dic_available_material["Material_capacity"], "Material_Constraint"
#Demand Constraint
problem += product_A <= 40, "Maximum_Product_A_Constraint"  
```

# Solver

```python
#Problem Solve
problem.solve()
```

```python
#Print the results
print(f"Status: {pl.LpStatus[problem.status]}")
print(f"Product A: {product_A.varValue}")
print(f"Product B: {product_B.varValue}")
print(f"Total Profit: ${pl.value(problem.objective)}")
```
